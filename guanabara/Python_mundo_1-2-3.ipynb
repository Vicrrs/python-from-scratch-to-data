#### Listas - Listas são mutáveis (podem ser alteradas) e ordenadas.

```python
frutas = ["macã", "banana", "laranja"]

print(frutas[0])
print(frutas[-1])

frutas.append("uva") # adiciona no final
print(frutas)

frutas.insert(1, "manga") #adicionando na posicao 1
print(frutas)

frutas.remove("banana") # remove pelo valor
print(frutas)

print(len(frutas)) # tamanho da lista
print("uva" in frutas) # true

print("-------Conteudo da lista-------")
for fruta in frutas:
    print(fruta)
```

```python
numeros = [5, 2, 9, 1, 7, 3]

print(14 * "-=-")


print("Fatiamento")
# Fatiamento (slicing)
print(numeros[:4])
print(numeros[1:4])
print(numeros[::-1]) # invertendo

print(14 * "-=-")

print("Ordenação")
# Ordenação
print(sorted(numeros)) # nova lista ordenada
numeros.sort(reverse=True) # ordena a propria lista
print(numeros)

print(14 * "-=-")

print("List Comprehension")

quadrados = [n ** 2 for n in range(1, 7)]
print(quadrados)

pares = [n for n in range(20) if n % 2 == 0]
print(pares)

print(14 * "-=-")

print("Enumerate e Zip")

nomes = ["Ana", "Bruno", "Carla"]
notas = [8.5, 7.0, 9.2]

for i, nome in enumerate(nomes, start=1):
    print(f"{i}. {nome}")

for nome, nota in zip(nomes, notas):
    print(f"{nome}: {nota}")

print(14 * "-=-")

print("Lista de listas (matriz)")

matriz = [[1, 2, 3], [4, 5, 6]]
print(matriz)
```

### FOR

O for é usado quando você sabe sobre o que vai iterar (uma lista, um intervalo, uma string etc.).

```python
numero = int(input("Digite um número: "))

for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

```python
precos = [12.5, 80.0, 45.9, 150.0, 9.99, 230.0]

total = 0
caros = []

for preco in precos:
    total += preco
    if preco > 100:
        caros.append(preco)

print(f"Total: R$ {total:.2f}")
print(f"Média: R$ {total / len(precos):.2f}")
print(f"Produtos acima de R$ 100: {caros}")
```

```python
alunos = {
    "Ana": [8, 9, 10],
    "Bruno": [5, 6, 4],
    "Carla": [7, 7, 8],
    "Diego": [3, 4, 5],
}

for posicao, (nome, notas) in enumerate(alunos.items(), start=1):
    media = sum(notas) / len(notas)
    situacao = "Aprovado" if media >= 6 else "Reprovado"
    print(f"{posicao}. {nome}: média {media:.1f} - {situacao}")

# Procurar o primeiro aluno reprovado e parar
for nome, notas in alunos.items():
    if sum(notas) / len(notas) < 6:
        print(f"\nPrimeiro reprovado encontrado: {nome}")
        break
```

### while

O while repete enquanto uma condição for verdadeira. É útil quando você não sabe de antemão quantas repetições serão necessárias.

```python
contador = 10

while contador > 0:
    print(contador)
    contador -= 1

print("Fogo!")
```

```python
senha_correta = "python123"
tentativas = 3

while tentativas > 0:
    senha = input("Digite a senha: ")

    if senha == senha_correta:
        print("Acesso liberado!")
        break
    else:
        tentativas -= 1
        print(f"Senha incorreta. Tentativas restantes: {tentativas}")

if tentativas == 0:
    print("Conta bloqueada!")
```

```python
import random

numero_secreto = random.randint(1, 100)
palpites = 0

print("Adivinhe o número entre 1 e 100!")

while True:
    entrada = input("Seu palpite (ou 'sair'): ")

    if entrada.lower() == "sair":
        print(f"Que pena! O número era {numero_secreto}.")
        break

    if not entrada.isdigit():
        print("Digite apenas números!")
        continue

    palpite = int(entrada)
    palpites += 1

    if palpite < numero_secreto:
        print("Mais alto! ⬆️")
    elif palpite > numero_secreto:
        print("Mais baixo! ⬇️")
    else:
        print(f"Acertou em {palpites} tentativas!")
        break
```

```python
# Exemplo 1: Área e perímetro do círculo (Matemática)

import math

def area_circulo(raio):
    return math.pi * raio ** 2

def perimetro_circulo(raio):
    return 2 * math.pi* raio

r = 5

print(f"Área: {area_circulo(r):.2f}")          # 78.54
print(f"Perímetro: {perimetro_circulo(r):.2f}") # 31.42
```

```python
# Exemplo 2: Fórmula de Bhaskara (Matemática)

import math

def bhaskara(a, b, c):
    """Resolve ax² + bx + c = 0 e retorna as raízes reais."""
    if a == 0:
        return "Não é uma equação de Segundo Grau."

    delta = b ** 2 - 4 * a * c

    if delta < 0:
        return "Não há raizer reais"
    elif delta == 0:
        x = -b / (2 * a)
        return (x,)
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        return (x1, x2)

print(bhaskara(1, -5, 6))   # (3.0, 2.0)
print(bhaskara(1, -4, 4))   # (2.0,)
print(bhaskara(1, 0, 1))    # Não há raízes reais
```

```python
# Exemplo 3: Movimento Uniformemente Variado (Física)
def posicao_mruv(s0, v0, a, t):
    """S = S0 + V0·t + (a·t²)/2"""
    return s0 + v0 * t + (a * t ** 2) / 2

def velocidade_mruv(v0, a, t):
    """V = V0 + a·t"""
    return v0 + a * t

# Um carro parte do repouso com aceleração de 2 m/s²
for t in range(0, 6):
    s = posicao_mruv(0, 0, 2, t)
    v = velocidade_mruv(0, 2, t)
    print(f"t = {t}s | posição = {s:.1f} m | velocidade = {v:.1f} m/s")
```

```python
# Exemplo 4: Queda livre com valor padrão (Física)

import math

def queda_livre(altura, g=9.81):
    """Calcula tempo de queda e velocidade final ao atingir o solo."""
    if altura < 0:
        raise ValueError("A altura não pode ser negativa")

    tempo = math.sqrt(2 * altura / g)
    velocidade = g * tempo
    return tempo, velocidade

# Na Terra (usa o g padrão)
t, v = queda_livre(45)
print(f"Terra -> tempo: {t:.2f}s | velocidade: {v:.2f} m/s")

# Na Lua (g = 1.62 m/s²)
t, v = queda_livre(45, g=1.62)
print(f"Lua   -> tempo: {t:.2f}s | velocidade: {v:.2f} m/s")
```

```python
# Exemplo 5: Energia mecânica e conservação (Física + Matemática)

def energia_cinetica(massa, velocidade):
    """Ec = (m·v²) / 2"""
    return massa * velocidade ** 2 / 2

def energia_potencial(massa, altura, g=9.81):
    """Ep = m·g·h"""
    return massa * g * altura

def energia_mecanica(massa, velocidade, altura, g=9.81):
    """Em = Ec + Ep"""
    return energia_cinetica(massa, velocidade) + energia_potencial(massa, altura, g)

def velocidade_no_solo(altura, v0=0, g=9.81):
    """Usa a conservação de energia: v = √(v0² + 2gh)"""
    return (v0 ** 2 + 2 * g * altura) ** 0.5

# Uma bola de 2 kg no alto de uma rampa de 10 m, parada
massa = 2
altura = 10

em_topo = energia_mecanica(massa, 0, altura)
v_final = velocidade_no_solo(altura)
em_solo = energia_mecanica(massa, v_final, 0)

print(f"Energia no topo: {em_topo:.2f} J")
print(f"Velocidade no solo: {v_final:.2f} m/s")
print(f"Energia no solo: {em_solo:.2f} J")
print(f"Energia conservada? {math.isclose(em_topo, em_solo)}" if (math := __import__('math')) else "")
```
