```python
# 1. Crie variáveis com seu nome, idade e cidade e imprima uma frase de apresentação usando f-string

nome = "Fulano"
idade = 24
cidade = "Gothan"

print(f"Olá eu sou {nome}, tenho {idade} anos de idade e sou de {cidade}")
```

```python
# 2. Leia dois números inteiros e imprima a soma, a diferença, o produto e a divisão entre eles.

num_01 = 10
num_02 = 20

soma = num_01 + num_02
subtracao = num_01 - num_02
multiplicacao = num_01 * num_02
divisao = num_01 / num_02

print(f"""
soma: {soma},\n
subtração: {subtracao}, \n
multiplicação: {multiplicacao}, \n
divisão: {divisao}
""")
```

```python
# 3. Leia o raio de um círculo (float) e imprima a área (use 3.14159 para pi).

raio = float(input("Informe o raio do círculo em cm: "))
pi = 3.14159

area = pi * raio

print(f"A área do círculo é: {area}")
```

```python
# 4. Leia um valor em reais e uma quantidade de produtos e imprima o total a pagar.

valor_produto = 10
qtd_produto = 56

valor_total = valor_produto * qtd_produto

print(valor_total)
```

```python
# 5 . Leia a temperatura em Celsius e converta para Fahrenheit: F = C * 9/5 + 32.

temp_Celsius = 35

temp_Fahrenheit = temp_Celsius * 9/5 + 32

print(f"A conversão de Celsius para Fahrenheit é {temp_Fahrenheit}")
```

```python
# 6. Leia um número de segundos e converta para o formato: X minutos e Y segundos (use // e %).

temp = int(input("Informe quantos segundos você ficou fora: "))

minutos = temp // 60
segundos = temp % 60

print(f"{minutos} minutos e {segundos} segundos")
```

```python
# 5. Leia dois números e troque os valores das variáveis, imprimindo antes e depois.
a = float(input("Informe o valor da variável a: "))
b = float(input("Informe o valor da variável b: "))

print(f"Os valores antes da troca: a = {a} e b = {b}")

a, b = b, a

print(f"Depois da troca: a = {a} e b = {b}")
```

```python
# 6. Leia um número de 3 dígitos e imprima cada dígito separadamente (dica: // e %).

num = int(input("Digite um número de 3 digitos: "))

unidade = num % 10
dezena = (num // 10) % 10
centena = num // 100

print("Unidade:", unidade)
print("Dezena:", dezena)
print("Centena:", centena)
```
